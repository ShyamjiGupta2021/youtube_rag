```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_ollama import ChatOllama
from langchain_core.prompts import PromptTemplate
from langchain_community.document_loaders import YoutubeLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_ollama import OllamaEmbeddings
from langchain_chroma import Chroma

llm = ChatOllama(model='llama3.1:8b', streaming=True)

loader = YoutubeLoader.from_youtube_url(youtube_url="https://www.youtube.com/live/S1xkJ7o1-JU?si=tG4k6e0Kq83zCUzc",language=["en","hi"])
load_doc = loader.load()

splitter= RecursiveCharacterTextSplitter(chunk_size= 1000, chunk_overlap=20)
split = splitter.split_text(load_doc[0].page_content)

embed = OllamaEmbeddings(model="nomic-embed-text:latest")
# embed_doc = embed.embed_documents(split)

store =Chroma.from_texts(texts=split,
                         embedding= embed)

retrive = store.as_retriever(search_kwargs={'k':5}, search_type='similarity')


prompt = PromptTemplate(
        template="""You are the teaching tutor and you have to give all the answers of the user query,
                    you are given the data, analyse the data and answer the query from the analysis, 
                    if you don't know the answer use your knowldge to answer the query!

                    data: {data}

                    query: {query}
        
                    """,
                    input_variables = ["data","query"]
    )



while True:  

    query = input()

    if query.lower() in ["bye", "exit"]:
        break
    
    ans = retrive.invoke(query)

    # print(ans)

    out_retrive = ""
    for i in ans:
        out_retrive += i.page_content + "\n\n"

    
    chain = prompt|llm 
    res=chain.stream({"data":out_retrive,"query":query})   

    for chunk in res:
        print(chunk.content, end="", flush=True)


    

```
